## Seaborn 시각화 카테고리 개요

이 노트북에서는 Seaborn의 주요 시각화 유형을 다룹니다. Seaborn은 Matplotlib을 기반으로 한 Python 데이터 시각화 라이브러리로, 통계적 그래프를 쉽게 그릴 수 있게 해줍니다.

### 주요 카테고리

| 카테고리 | 설명 | 주요 함수 |
|---|---|---|
| **분포(Distribution)** | 데이터의 분포 형태를 파악 | `histplot`, `kdeplot`, `ecdfplot`, `rugplot`, `displot` |
| **관계형(Relational)** | 두 변수 간의 관계를 시각화 | `scatterplot`, `lineplot`, `relplot` |
| **범주형(Categorical)** | 범주별 데이터의 분포나 값을 비교 | `barplot`, `countplot`, `boxplot`, `violinplot`, `stripplot`, `swarmplot`, `pointplot`, `catplot` |
| **회귀(Regression)** | 회귀 직선과 잔차 분석 | `regplot`, `lmplot`, `residplot` |
| **행렬(Matrix)** | 데이터 행렬의 패턴을 색상으로 표현 | `heatmap`, `clustermap` |
| **다변량(Multivariate)** | 여러 변수 간의 관계를 한 번에 시각화 | `pairplot`, `PairGrid`, `FacetGrid`, `jointplot` |

각 카테고리는 데이터의 특성과 분석 목적에 따라 적절히 선택하여 사용합니다.


````python
# 분포(Distribution) — histplot, kdeplot, ecdfplot, rugplot, displot
# 관계형(Relational) — scatterplot, lineplot, relplot
# 범주형(Categorical) — barplot, countplot, boxplot, violinplot, stripplot, swarmplot, pointplot, catplot
# 회귀(Regression) — regplot, lmplot, residplot
# 행렬(Matrix) — heatmap, clustermap
# 다변량(Multivariate) — pairplot, PairGrid, FacetGrid, jointplot
````

### 라이브러리 임포트

데이터 시각화에 필요한 핵심 라이브러리를 임포트합니다.

- **`matplotlib.pyplot`**: Python의 기본 시각화 라이브러리인 Matplotlib의 pyplot 모듈입니다. `plt`라는 별칭으로 주로 사용되며, 그래프의 기본 틀(figure, axes)을 생성하고 제어하는 역할을 합니다.
- **`seaborn`**: 통계적 데이터 시각화를 위한 고수준 인터페이스를 제공하는 라이브러리입니다. `sns`라는 별칭으로 사용되며, Matplotlib 기반으로 더 아름답고 통계적으로 의미 있는 그래프를 쉽게 그릴 수 있게 해줍니다.
- **`pandas`**: 데이터 조작 및 분석을 위한 핵심 라이브러리입니다. DataFrame이라는 2차원 테이블 형태의 데이터 구조를 제공하며, Seaborn의 `data` 파라미터에 주로 전달됩니다.

이 세 가지 라이브러리는 Python 데이터 과학 생태계에서 시각화 작업의 표준 조합입니다.


````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
````

## 범주형(Categorical) — barplot

### barplot 개념

`barplot`은 범주형 변수(x)에 따른 수치형 변수(y)의 **집계 값(기본적으로 평균)**을 막대 형태로 표현하는 그래프입니다.

### 핵심 파라미터

- **`data`**: 그래프를 그릴 DataFrame 객체
- **`x`**: 범주형 변수(가로축) — 여기서는 `"day"`(요일: Thur, Fri, Sat, Sun)
- **`y`**: 수치형 변수(세로축) — 여기서는 `"tip"`(팁 금액)

### 작동 원리

1. x축의 각 범주(요일)별로 데이터를 그룹화합니다.
2. 각 그룹의 y값(`tip`)에 대한 **평균(mean)**을 계산합니다.
3. 계산된 평균값이 막대의 높이가 됩니다.
4. **에러바(errorbar)**가 자동으로 표시되어 각 그룹 평균의 신뢰구간(95% CI)을 나타냅니다.

### 에러바(Error Bar)의 의미

- 막대의 중간 높이가 해당 범주의 평균값입니다.
- 에러바의 길이가 길수록 해당 범주 내 데이터의 변동(분산)이 크다는 것을 의미합니다.
- 에러바가 짧으면 데이터가 평균 주변에 밀집해 있어 평균값이 믿을만하다는 의미입니다.

> **참고**: `errorbar=None`을 지정하면 에러바를 제거할 수 있습니다. 기본값은 `errorbar=("ci", 95)`입니다.


````python
# 범주형(Categorical) — barplot

# 기본 barplot
sns.barplot(data=df_tips, x="day", y="tip") # 요일 별 팁의 평균을 막대 형태로 그리는 것
plt.show()

# errorbar=None이 디폴트
# sns.barplot(data=df_tips, x="day", y="tip", errorbar=None)
# errorbar의 중간이 막대의 높이(평균)
# errorbar의 길이가 길수록 해당 category의 데이터 변동이 크다는 것을 의미
````

### barplot — 신뢰구간(Confidence Interval) 조정

`errorbar` 파라미터를 사용하여 평균값 주변의 신뢰구간(CI, Confidence Interval)을 조정할 수 있습니다.

- **`errorbar=("ci", 95)`**: 95% 신뢰구간 (기본값)
  - 평균값이 95%의 확률로 이 범위 안에 존재한다는 통계적 의미입니다.
  - 신뢰구간 값이 커지면(예: 99%) 에러바는 더 길어집니다.
  - 이는 우리가 더 높은 확신을 가지려 할수록 구간이 넓어지는 통계적 원리를 반영합니다.

- **`errorbar=("ci", 99)`**: 99% 신뢰구간
  - 95%보다 더 넓은 범위를 나타냅니다.
  - 데이터의 변동성이 클수록 신뢰구간도 넓어집니다.

### 해석 방법

막대의 높이는 평균이며, 실제 데이터 포인트들은 막대 위아래(에러바 범위)에 분포합니다. 에러바가 겹치는 두 범주는 통계적으로 유의미한 차이가 없을 가능성이 있습니다.


````python

# errorbar=("ci", 95) → 95% 신뢰구간 (기본값)
# 평균값이 95% 확률로 이 범위 안에 있음을 의미
# 확률값이 커지면 에러바의 크기도 커짐
# 막대 높이는 평균이고 실제 데이터는 그 위아래에 존재함
sns.barplot(data=df_tips, x="day", y="tip", errorbar=("ci", 95))
# sns.barplot(data=df_tips, x="day", y="tip", errorbar=("ci", 99))
````

### barplot — 표준오차(Standard Error) 사용

`errorbar` 파라미터에 `"se"`(Standard Error)를 지정하여 표준오차를 에러바로 사용할 수 있습니다.

### 표준오차 vs 표준편차

- **표준편차(Standard Deviation, SD)**: 데이터가 평균으로부터 흩어진 정도를 나타냅니다. 전체 데이터의 변동성을 측정합니다.
- **표준오차(Standard Error, SE)**: 평균 추정치의 정확도를 나타냅니다. 공식은 **SE = SD / √n**입니다.

### 해석

- 표준편차가 크면 표준오차도 커집니다.
- 데이터 개수(n)가 많아지면 표준오차는 작아집니다(√n으로 나누기 때문).
- 표준오차는 **'평균을 얼마나 믿을 수 있는가?'**를 보여주는 지표입니다.
- 표본 크기가 클수록 표준오차가 작아져 평균 추정치가 더 정확해집니다.

> **참고**: `errorbar=("se", 1)`에서 1은 표준오차의 배수를 의미합니다. 1이면 1SE, 2이면 2SE를 의미합니다.


````python
# errorbar=("se", 1) → 표준오차
# 표준오차 = 표준편차 / √n
# 표준편차가 크면 표준오차는 커지고, 데이터 개수가 크면 작아짐 → 평균을 얼마나 믿을 수 있는가?
sns.barplot(data=df_tips, x="day", y="tip", errorbar=("se", 1))
````

### barplot — hue와 palette

범주형 변수를 추가로 구분하여 막대를 더 세분화할 수 있습니다.

- **`hue`**: 추가적인 범주형 변수를 지정하여 각 x축 범주 내에서 hue에 따라 막대를 분할합니다.
  - 예: `hue="sex"`는 각 요일별로 남성(Male)과 여성(Female)의 팁 평균을 나란히 보여줍니다.
- **`palette`**: 막대의 색상 조합을 지정합니다.
  - `"Set2"`: 부드러운 색조의 팔레트로, 색각 이상을 가진 사람들도 구분하기 쉬운 색상 조합입니다.
  - 기타 팔레트: `"deep"`, `"muted"`, `"pastel"`, `"bright"`, `"dark"`, `"colorblind"` 등

### 해석

hue를 사용하면 같은 x축 범주 내에서 서로 다른 하위 그룹 간의 비교가 가능해집니다. 예를 들어, '토요일에 남성과 여성 중 누가 더 많은 팁을 주는가?'와 같은 질문에 답할 수 있습니다.


````python
# 범주형(Categorical) — barplot

# hue + palette
sns.barplot(data=df_tips, x="day", y="tip", hue="sex", palette="Set2")
plt.show()
````

### barplot — estimator와 errorbar

`estimator` 파라미터를 사용하여 막대의 높이를 결정하는 집계 함수를 변경할 수 있습니다.

### 지원하는 Estimator

| Estimator | 설명 |
|---|---|
| `"mean"` | 평균 (기본값) |
| `"sum"` | 합계 |
| `"max"` | 최댓값 |
| `"min"` | 최솟값 |
| `"median"` | 중앙값 |

### 주의사항

- `estimator="sum"`, `"max"`, `"min"` 등 평균이 아닌 estimator를 사용할 때는 `errorbar=None`으로 설정하는 것이 일반적입니다.
- 왜냐하면 신뢰구간이나 표준오차는 평균 추정치의 불확실성을 나타내는 개념이므로, 합계나 최댓값에는 통계적으로 적절하지 않기 때문입니다.
- `errorbar=None`을 지정하면 막대만 표시되어 그래프가 더 깔끔해집니다.

### 해석

`estimator="sum"`을 사용하면 각 요일의 **총 팁 수입**을 알 수 있습니다. 이는 평균 팁과는 다른 비즈니스 관점의 인사이트를 제공합니다.


````python
# 범주형(Categorical) — barplot

# estimator + errorbar
# estimator는 막대 높이를 무엇으로 결정할 것인지
# "mean", "sum", "max", "min", "median"
sns.barplot(data=df_tips, x="day", y="tip", estimator="sum", errorbar=None) # errorbar는 sum, max, min의 경우 의미가 없음
plt.show()
````

### barplot — 순서 지정 (order, hue_order)

범주형 변수의 표시 순서를 직접 제어할 수 있습니다.

- **`order`**: x축에 표시되는 범주의 순서를 리스트로 지정합니다.
  - 기본적으로는 범주가 알파벳 순서로 정렬됩니다.
  - 예: `order=["Thur", "Fri", "Sat", "Sun"]`는 요일을 시간 순서대로 정렬합니다.
- **`hue_order`**: hue 범주 내의 순서를 지정합니다.
  - 예: `hue_order=["Male", "Female"]`는 남성을 왼쪽, 여성을 오른쪽에 배치합니다.
  - `hue_order=["Female", "Male"]`로 바꾸면 순서가 반전됩니다.

### 활용

순서 지정은 데이터의 논리적 흐름(시간 순서, 중요도 순서 등)을 반영하여 그래프를 더 직관적으로 만듭니다. 특히 요일이나 월처럼 자연스러운 순서가 있는 범주에서 유용합니다.


````python
# 범주형(Categorical) — barplot

# order + hue_order
sns.barplot(data=df_tips, x="day", y="tip",
            hue="sex",
            order=["Thur", "Fri", "Sat", "Sun"],    # x축에 표시되는 범주의 순서를 직접 지정
            hue_order=["Male", "Female"])           # 요일 내 순서
            # hue_order=["Female", "Male"])
plt.show()
````

### barplot — 방향 및 스타일 (orient, width, alpha)

막대의 방향과 시각적 스타일을 조정할 수 있습니다.

- **`orient="h"`**: 가로 막대 그래프로 변경합니다.
  - x축에 수치형 변수(`tip`), y축에 범주형 변수(`day`)를 배치합니다.
  - 범주 레이블이 길 때 유용합니다.
- **`width`**: 막대의 너비를 조정합니다. (0~1 사이 값, 1이 최대)
  - `width=0.9`: 기본값에 가까운 너비
  - `width=0.5`: 더 얇은 막대, 범주 간 간격이 넓어짐
- **`alpha`**: 막대의 투명도를 조정합니다. (0~1 사이 값)
  - `alpha=0.7`: 70% 불투명도
  - 여러 그래프를 겹쳐 그릴 때 유용합니다.

### 가로 vs 세로 막대

- 세로 막대(`orient="v"`, 기본): 범주가 x축, 값이 y축 — 일반적인 경우
- 가로 막대(`orient="h"`): 범주가 y축, 값이 x축 — 범주 레이블이 길거나 범주가 많을 때 추천


````python
# 범주형(Categorical) — barplot

# orient + width + alpha
sns.barplot(data=df_tips, x="tip", y="day", orient="h", width=0.9, alpha=0.7) # width는 막대의 너비 (요일 방향이며 1이 최대값)
plt.show()
````

## 범주형(Categorical) — countplot

### countplot 개념

`countplot`은 각 범주에 속한 **데이터의 개수(빈도)**를 막대 형태로 표현하는 그래프입니다.

### barplot과의 차이

| 구분 | barplot | countplot |
|---|---|---|
| y축 의미 | 수치형 변수의 집계값(평균 등) | 각 범주의 데이터 개수 |
| y 파라미터 | 필요함 | 필요없음 (자동 계산) |
| 용도 | 범주별 수치 비교 | 범주별 빈도/비율 비교 |

### 핵심 파라미터

- **`data`**: DataFrame
- **`x`**: 세어 볼 범주형 변수 — 여기서는 `"class"`(타이타닉 객실 등급: First, Second, Third)

### 해석

타이타닉 데이터셋에서 `countplot(data=df_titanic, x="class")`는 각 객실 등급(1등석, 2등석, 3등석)에 탑승한 승객 수를 보여줍니다. 이를 통해 '어느 등급에 가장 많은 사람이 탔는가?'를 한눈에 파악할 수 있습니다.


````python
# 범주형(Categorical) — countplot

# 기본 countplot
sns.countplot(data=df_titanic, x="class")
plt.show()
````

### countplot — hue와 palette

`barplot`과 마찬가지로 `hue`를 사용하여 추가 범주로 막대를 분할할 수 있습니다.

- **`hue="survived"`**: 각 객실 등급 내에서 생존자(1)와 사망자(0)의 수를 나란히 비교합니다.
- **`palette="Set2"`**: 부드러운 색상 조합을 적용합니다.

### 해석

이 그래프를 통해 '객실 등급별 생존/사망 분포는 어떠한가?'라는 질문에 답할 수 있습니다.
예를 들어, 1등석 승객 중 생존자가 사망자보다 많고, 3등석은 그 반대일 가능성을 시각적으로 확인할 수 있습니다.
이는 '구명정 탑승 시 객실 등급이 영향을 미쳤는가?'라는 역사적 질문에 대한 시각적 증거를 제공합니다.


````python
# 범주형(Categorical) — countplot

# hue + palette
sns.countplot(data=df_titanic, x="class", hue="survived", palette="Set2")
plt.show()
````

### countplot — 순서 지정 (order, hue_order)

`barplot`과 동일하게 `order`와 `hue_order`를 사용할 수 있습니다.

- **`order=["First", "Second", "Third"]`**: 객실 등급을 1등석 → 2등석 → 3등석 순서로 정렬합니다.
- **`hue_order=[1, 0]`**: survived 컬럼의 순서를 생존(1) → 사망(0)으로 지정합니다.
  - `survived` 컬럼은 원래 0(사망)과 1(생존)의 정수 값을 가집니다.
  - `hue_order`를 지정하지 않으면 기본적으로 0 → 1 순서(알파벳/숫자 오름차순)로 표시됩니다.

### 활용

순서 지정은 '등급이 높을수록 생존율이 높은가?'와 같은 가설을 시각적으로 검증할 때 특히 유용합니다.


````python
# 범주형(Categorical) — countplot

sns.countplot(data=df_titanic, x="class", hue="survived",
              order=["First", "Second", "Third"],
              hue_order=[1, 0]) # survived 컬럼은 0(사망), 1(생존) 숫자
````

### countplot — 방향 및 너비 (orient, width)

`barplot`과 동일하게 가로 방향과 막대 너비를 조정할 수 있습니다.

  - 기본값은 약 0.8입니다.
  - 값이 작을수록 막대가 얇아지고 범주 간 간격이 넓어집니다.

### 활용

가로 countplot은 범주 레이블이 길 때(예: 국가명, 제품명) 가독성을 높여줍니다.


````python
# 범주형(Categorical) — countplot

# orient + width
sns.countplot(data=df_titanic, y="class", orient="h", width=0.4)
plt.show()
````

### boxplot — 이상치 마커 스타일 (flierprops)

`flierprops` 파라미터를 사용하여 이상치(outlier) 데이터 포인트의 시각적 스타일을 커스터마이징할 수 있습니다.

### flierprops 옵션

- **`marker`**: 마커 모양
  - `"x"`: X 모양
  - `"o"`: 원형 (기본값)
  - `"s"`, `"D"`, `"^"` 등 다양한 모양 가능
- **`markeredgecolor`**: 마커 테두리 색상 (채움이 없는 마커용)
  - 예: `"red"` — 빨간색 테두리
- **`markerfacecolor`**: 마커 내부 채움 색상 (채움이 있는 마커용)
  - 예: `"red"` — 빨간색 채움
- **`markersize`**: 마커 크기 (기본값: 5)

### 스타일 조합

- **채움이 없는 마커** (`"x"`, `"+"`): `markeredgecolor`로 색상 지정
- **채움이 있는 마커** (`"o"`, `"s"`): `markerfacecolor`로 색상 지정

### 활용

이상치를 더 눈에 띄게 만들어 데이터 정제나 이상 탐지 작업에서 유용합니다.


````python
# 범주형(Categorical) — boxplot

# 이상치 마커 스타일
# "x" 마커 → markeredgecolor로 색상 지정
sns.boxplot(data=df_penguins, x="species", y="flipper_length_mm",
            flierprops={"marker": "x", "markeredgecolor": "red", "markersize": 8})

# 채움이 있는 마커 → markerfacecolor 사용 가능
sns.boxplot(data=df_penguins, x="species", y="flipper_length_mm",
            flierprops={"marker": "o", "markerfacecolor": "red", "markersize": 8})

plt.show()
````

### boxplot 이론 — 사분위수와 이상치

박스플롯을 완전히 이해하려면 사분위수(Quartile)와 이상치 판정 기준을 알아야 합니다.

### 사분위수 (Quartiles)

| 기호 | 이름 | 백분위 | 의미 |
|---|---|---|---|
| Q1 | 제1사분위수 | 하위 25% | 데이터의 25%가 이 값보다 작음 |
| Q2 | 제2사분위수(중앙값) | 하위 50% | 데이터의 절반이 이 값보다 작음 |
| Q3 | 제3사분위수 | 하위 75% | 데이터의 75%가 이 값보다 작음 |
| Q4 | 최댓값 | 100% | 데이터 중 가장 큰 값 |

### 박스플롯 구성 요소

- **박스 아래 경계**: Q1
- **박스 중앙 선**: Q2 (중앙값, Median)
- **박스 위 경계**: Q3
- **수염(whisker) 하단**: Q1 - 1.5×IQR (단, 데이터의 최솟값보다 작지 않음)
- **수염(whisker) 상단**: Q3 + 1.5×IQR (단, 데이터의 최댓값보다 크지 않음)
- **이상치(outlier)**: 수염 범위를 벗어난 점들

### IQR (InterQuartile Range, 사분위 범위)

- **IQR = Q3 - Q1**
- 데이터의 중간 50%가 분포하는 범위입니다.
- IQR이 크면 데이터가 중앙값 주변에 넓게 퍼져 있다는 의미입니다.

### 기타 분위수 용어

| 용어 | 등분 | 기준점 개수 |
|---|---|---|
| Tertile | 3등분 | 2개 |
| Quartile | 4등분 | 3개 |
| Quintile | 5등분 | 4개 |
| Percentile | 100등분 | 99개 |

### 이상치 판정 기준

- **Q1 - 1.5×IQR보다 작은 값**: 하위 이상치
- **Q3 + 1.5×IQR보다 큰 값**: 상위 이상치
- 1.5라는 계수는 Tukey(1977)가 제안한 경험적 기준입니다.
- 더 엄격하게 판정하려면 3.0을 사용할 수 있습니다(극단 이상치, extreme outlier).


````python
# Q1        하위 25%        데이터의 25%가 이 값보다 작음
# Q2        하위 50%        중앙값, 데이터의 절반이 이 값보다 작음
# Q3        하위 75%        데이터의 75%가 이 값보다 작음
# Q4        100%            최댓값

# 박스플롯에서 박스의 아래가 Q1, 위가 Q3, 중앙선이 Q2

# Tertile       3등분       →  2개의 기준점
# Quartile      4등분       →  3개의 기준점
# Quintile      5등분       →  4개의 기준점
# Percentile    100등분     → 99개의 기준점

# IQR: InterQuartile Range이며 Q1, Q2, Q3를 의미
# IQR(사분위 범위) = Q3 - Q1

# 박스폴롯의 하단 및 상단
# Q1 - 1.5×IQR
# Q3 + 1.5×IQR

# Q1 - 1.5×IQR보다 작거나 Q3 + 1.5×IQR보다 큰 값은 이상치
````

## 범주형(Categorical) — violinplot

### violinplot 개념

`violinplot`은 박스플롯과 KDE(커널 밀도 추정)를 결합한 그래프입니다.
데이터의 분포 형태(비대칭성, 다봉 분포 등)를 시각적으로 파악할 수 있습니다.

### 구성 요소

- **외부 곡선**: KDE(커널 밀도 추정) 곡선
  - 데이터가 많이 몰린 곳은 곡선이 넓어집니다.
  - 데이터가 적은 곳은 곡선이 좁아집니다.
- **내부 표시**: `inner` 파라미터로 결정 (기본값은 `"box"`)

### inner 파라미터 옵션

| 값 | 설명 |
|---|---|
| `"box"` | 기본값. 박스플롯 형태(Q1, Q2, Q3)를 내부에 표시 |
| `"quart"` | 사분위수 가로선만 표시 |
| `"point"` | 개별 데이터 포인트를 점으로 표시 |
| `"stick"` | 개별 데이터를 세로선으로 표시 |
| `None` | 내부 표시 없음 (KDE 곡선만) |

### 박스플롯과의 비교

- **박스플롯**: 요약 통계량(Q1, Q2, Q3, 이상치)만 보여줌
- **바이올린플롯**: 전체 분포의 형태(밀도)를 보여줌
- 바이올린플롯이 더 많은 정보를 제공하지만, 정확한 수치는 박스플롯이 더 읽기 쉬움


````python
# 범주형(Categorical) — violinplot

# 외부 곡선 → KDE (커널 밀도 추정) 곡선, 데이터가 많이 몰린 곳이 넓어짐
# 내부     → inner 파라미터로 결정(디폴트는 "box")

# "box"         기본, 박스플롯 (Q1, Q2, Q3)
# "quart        "Q1, Q2, Q3 가로선만
# "point        "개별 데이터 점
# "stick"       개별 데이터 세로선
# None          아무것도 표시 안 함

# 기본 violinplot
sns.violinplot(data=df_penguins, x="species", y="flipper_length_mm")
plt.show()
````

### violinplot — hue와 split

`hue`를 사용하여 추가 범주로 분리하고, `split=True`로 좌우로 나누어 표시할 수 있습니다.

- **`hue="sex"`**: 성별(Male/Female)로 데이터를 나눕니다.
- **`split=True`**: hue의 두 범주를 하나의 바이올린을 좌우로 분할하여 표시합니다.
  - 왼쪽: hue의 첫 번째 범주(기본적으로 알파벳 순)
  - 오른쪽: hue의 두 번째 범주
- **`palette="Set2"`**: 색상 팔레트 적용

### split의 장점

- `split=True`를 사용하면 두 범주의 분포를 동일한 x축 위치에서 직접 비교할 수 있습니다.
- `split=False`(기본)일 경우 hue에 따라 별도의 바이올린이 생성되어 공간을 더 차지합니다.
- 분포의 모양 차이(예: 남성은 정규분포, 여성은 비대칭)를 한눈에 파악할 수 있습니다.


````python
# 범주형(Categorical) — violinplot

# hue + split — hue 2개일 때 좌우로 분리
sns.violinplot(data=df_penguins, x="species", y="flipper_length_mm",
               hue="sex", split=True, palette="Set2")
plt.show()
````

### violinplot — inner 옵션 상세

`inner` 파라미터를 변경하여 바이올린 내부의 표시 방식을 다양하게 조정할 수 있습니다.

- **`inner="quart"`**: Q1, Q2, Q3를 가로선으로만 표시
  - 박스플롯보다 미니멀한 느낌
  - KDE 곡선과 함께 사분위수 위치를 보여줌
- **`inner="point"`**: 각 데이터 포인트를 점으로 표시
  - 데이터 개수가 적을 때 유용
- **`inner="stick"`**: 각 데이터를 세로선으로 표시
  - 'rug plot'과 유사한 효과
- **`inner=None`**: 내부 표시 제거
  - KDE 곡선의 아름다움을 살리고 싶을 때
  - 다른 그래프(예: stripplot)와 겹쳐 그릴 때 깔끔함

### 선택 가이드

| 데이터 특성 | 추천 inner |
|---|---|
| 요약 통계량이 중요 | `"box"` 또는 `"quart"` |
| 데이터 개수가 적음 | `"point"` 또는 `"stick"` |
| 다른 플롯과 겹침 | `None` |


````python
# 범주형(Categorical) — violinplot

# inner — 바이올린 안쪽 표시 방식
# "box" (기본) / "quart" / "point" / "stick" / None
sns.violinplot(data=df_penguins, x="species", y="flipper_length_mm",
               inner="quart")
plt.show()
````

## 범주형(Categorical) — stripplot과 swarmplot

### 개념 비교

두 함수는 모두 범주형 변수에 따른 **원본 데이터 포인트**를 직접 점으로 찍어 보여줍니다.

| 구분 | stripplot | swarmplot |
|---|---|---|
| **겹침** | 점들이 겹칠 수 있음 | 점들이 겹치지 않음 |
| **알고리즘** | 무작위(random) 좌우 흔들기 | 알고리즘적 배치 |
| **데이터 손실** | 없음 | 없음 |
| **시각적 특성** | `jitter`로 흩어보임 | 꿀벌 떼(swarm)처럼 정렬됨 |
| **속도** | 빠름 | 데이터가 많을 때 느림 |
| **용도** | 대용량 데이터 | 중소용량 데이터의 정밀 분포 확인 |

### stripplot의 jitter

- **`jitter=True`**: x축 범주 내에서 점들을 무작위로 좌우로 흔듭니다.
  - 기본값입니다.
  - 같은 위치에 여러 점이 겹치는 것을 방지합니다.
- **`jitter=False`**: 모든 점을 중앙에 정렬합니다.
  - 데이터가 많으면 점들이 완전히 겹쳐 보입니다.
- **`jitter=0.2`**: 흔들림 정도를 0~1 사이 값으로 조절합니다.

### swarmplot의 특징

- 점들이 겹치지 않도록 알고리즘이 최적의 위치를 계산합니다.
- 데이터의 실제 분포 밀도를 더 정확하게 시각화합니다.
- 데이터 개수가 많으면 계산 시간이 길어질 수 있습니다.


````python
# stripplot (jitter=True) → 랜덤하게 좌우로 흔들어서 배치 (겹칠 수 있음); random
# swarmplot               → 겹치지 않도록 알고리즘이 좌우로 배치; algorithmic
````

### stripplot — 기본 사용법

`stripplot`은 범주형 변수에 따른 원본 데이터 포인트를 직접 표시합니다.

### 핵심 파라미터

- **`data`**: DataFrame
- **`x`**: 범주형 변수 — 여기서는 `"day"`(요일)
- **`y`**: 수치형 변수 — 여기서는 `"tip"`(팁 금액)

### 해석

각 요일별로 개별 팁 데이터가 점으로 표시됩니다.
점의 수 = 해당 요일의 데이터 개수입니다.
점들이 모여 있는 곳은 데이터가 많이 분포하는 영역이며,
높은 위치의 점은 큰 팁을 의미합니다.

### 장점과 단점

- **장점**: 원본 데이터를 직접 볼 수 있어 데이터의 실제 분포를 파악할 수 있습니다.
- **단점**: 데이터가 많으면 점들이 겹쳐서 개별 포인트를 구분하기 어렵습니다.
  - 이럴 때 `jitter`를 조정하거나 `swarmplot`을 사용하면 됩니다.


````python
# 범주형(Categorical) — stripplot

# 기본 stripplot
sns.stripplot(data=df_tips, x="day", y="tip")
plt.show()
````

- **`hue="sex"`**: 남성과 여성을 다른 색으로 표시합니다.
- **`palette="Set2"`**: 색상 팔레트 적용

### 해석

각 요일의 팁 데이터를 성별로 색상 구분하여 표시합니다.
이를 통해 '특정 요일에 남성/여성 중 누가 더 큰 팁을 주는가?'를
개별 데이터 수준에서 확인할 수 있습니다.

> **팁**: stripplot은 `boxplot`이나 `violinplot`과 함께 사용하면
> 요약 통계량과 원본 데이터를 동시에 볼 수 있어 더 풍부한 인사이트를 얻을 수 있습니다.

### stripplot — jitter 조정

`jitter` 파라미터를 사용하여 점의 좌우 흔들림 정도를 제어합니다.

- **`jitter=True`**: 기본값. 적절한 정도로 무작위 흔들기를 적용합니다.
- **`jitter=False`**: 흔들기를 비활성화합니다.
  - 모든 점이 범주 중앙에 정렬되어 겹침 현상이 심해집니다.
  - 데이터가 매우 적을 때나 의도적으로 겹침을 보여주고 싶을 때 사용합니다.
- **`jitter=0.2`**: 흔들림 정도를 0~1 사이의 값으로 조절합니다.
  - 값이 클수록 점들이 더 넓게 흩어집니다.
  - 값이 작을수록 점들이 중앙에 가깝게 모입니다.

### 활용

`jitter` 값을 조절하면 데이터 밀도와 개별 포인트 가시성 사이의 균형을 맞출 수 있습니다.


````python
# 범주형(Categorical) — stripplot

# jitter — 점 겹침 방지를 위한 좌우 흔들기
# True (기본) / False / 0.0~1.0 (흔들기 정도)
sns.stripplot(data=df_tips, x="day", y="tip", jitter=False)
# sns.stripplot(data=df_tips, x="day", y="tip", jitter=0.2)
plt.show()
````

### swarmplot — 기본 사용법

`swarmplot`은 점들이 겹치지 않도록 알고리즘적으로 배치하는 stripplot의 변형입니다.

### 핵심 특징

- 점들이 겹치지 않도록 좌우로 배치됩니다.
- 데이터의 실제 분포 형태를 더 정확하게 시각화합니다.
- 점들이 '꿀벌 떼(swarm)'처럼 모여 있는 모양을 보여줍니다.

### 해석

stripplot과 달리 데이터가 많은 영역이 시각적으로 더 넓게 표현됩니다.
이는 데이터의 밀도(density)를 더 정확하게 직관적으로 전달합니다.
하지만 데이터 개수가 매우 많으면(수천 개 이상) 계산 시간이 길어지고
점들이 너무 작아져 가독성이 떨어질 수 있습니다.

### stripplot과의 선택 기준

| 상황 | 추천 |
|---|---|
| 데이터 < 500개 | `swarmplot` |
| 데이터 > 1000개 | `stripplot` (적절한 jitter 사용) |
| 밀도 정확도가 중요 | `swarmplot` |
| 속도가 중요 | `stripplot` |


````python
# 범주형(Categorical) — swarmplot

# 기본 swarmplot
sns.swarmplot(data=df_tips, x="day", y="tip")
plt.show()
````

### swarmplot — hue와 palette

`swarmplot`도 `hue`와 `palette`를 지원하여 추가 범주로 점들을 색상 구분할 수 있습니다.

- **`hue="sex"`**: 남성과 여성을 다른 색으로 구분합니다.
- **`palette="Set2"`**: 색상 팔레트 적용

### 해석

성별에 따른 팁 분포를 개별 데이터 수준에서 색상으로 구분하여 볼 수 있습니다.
특정 요일/성별 조합에서 데이터가 어떻게 분포하는지 정확하게 파악할 수 있습니다.

> **참고**: `swarmplot`에 `hue`를 사용하면 자동으로 점들이 hue별로 그룹화되어 배치됩니다.


````python
# 범주형(Categorical) — swarmplot

# hue + palette
sns.swarmplot(data=df_tips, x="day", y="tip", hue="sex", palette="Set2")
plt.show()
````

### stripplot + swarmplot 조합

두 플롯을 겹쳐서 사용하면 더 풍부한 시각적 효과를 얻을 수 있습니다.

- **`sns.stripplot(..., alpha=0.3)`**: 원본 데이터를 투명하게 표시하여 전체 분포의 윤곽을 보여줍니다.
- **`sns.swarmplot(..., color="black", size=4)`**: 데이터 포인트를 검은색으로 선명하게 표시합니다.

### 효과

- stripplot의 반투명한 점들은 데이터의 전체적인 분포 밀도를 보여줍니다.
- swarmplot의 검은 점들은 개별 데이터 포인트를 명확하게 보여줍니다.
- 이 조합은 '데이터가 전체적으로 어디에 몰려 있고, 개별 포인트는 어디에 있는가?'를 동시에 전달합니다.

### 활용

발표용 그래프나 보고서에서 데이터의 전체적 맥락과 개별 사례를 동시에 보여주고 싶을 때 유용합니다.


````python
# 범주형(Categorical) — swarmplot

# stripplot + swarmplot 조합
sns.stripplot(data=df_tips, x="day", y="tip", alpha=0.3)
sns.swarmplot(data=df_tips, x="day", y="tip", color="black", size=4)
plt.show()
````

## 다변량(Multivariate) — pairplot

### pairplot 개념

`pairplot`은 데이터셋의 **수치형 컬럼들 간의 모든 조합**을 한 번에 시각화하는 그래프입니다.
산점도 행렬(Scatter Plot Matrix)이라고도 불립니다.

### 격자 구조

컬럼이 3개(A, B, C)인 경우, 3×3 격자가 생성됩니다:

```
      A         B         C
A [대각선]   [A vs B]  [A vs C]
B [B vs A]  [대각선]   [B vs C]
C [C vs A]  [C vs B]  [대각선]
```

- **대각선(diag_kind)**: 각 컬럼의 **단변량 분포**를 표시
  - `"hist"`: 히스토그램 (기본값, hue가 없을 때)
  - `"kde"`: 커널 밀도 추정 곡선 (hue가 있을 때 기본값)
- **비대각선(kind)**: 두 컬럼 간의 **이변량 관계**를 표시
  - `"scatter"`: 산점도 (기본값)
  - `"kde"`: 2D KDE 등고선
  - `"hist"`: 2D 히스토그램
  - `"reg"`: 회귀선이 추가된 산점도

### hue의 효과

- `hue`를 지정하면 대각선 성분이 자동으로 KDE로 변경됩니다.
- 비대각선 산점도의 점들이 hue 범주별로 색상 구분됩니다.

### 활용

수치형 변수가 여러 개일 때, 변수들 간의 **상관관계와 분포를 전체적으로 파악**하는 데 필수적인 도구입니다.
특히 머신러닝의 특성 공학(Feature Engineering) 단계에서 유용합니다.


````python
# 다변량(Multivariate) — pairplot
# 수치형 컬럼들의 모든 조합을 격자 형태로 한번에 그려주는 그래프

# 컬럼이 3개 (A, B, C)인 경우 3×3 격자 생성

#      A         B         C
# A [대각선]   [A vs B]  [A vs C]
# B [B vs A]  [대각선]   [B vs C]
# C [C vs A]  [C vs B]  [대각선]

# 대각선(diag_kind): 히스토그램 (diag_kind="hist") or KDE (diag_kind="kde") (각 컬럼의 단변량 분포)
# hue를 사용하면 대각성분은 KDE, hue가 없으면 대각성분은 hist
# 비대각선(kind): 산점도 (두 컬럼 간의 관계)

df_penguins.dtypes
````

### pairplot — hue와 palette

`pairplot`에 `hue`를 추가하면 범주형 변수에 따라 점들을 색상 구분하여 관계를 더 세밀하게 분석할 수 있습니다.

### 해석

- **대각선**: 각 종별로 KDE 곡선이 겹쳐 표시됩니다.
  - 예: 세 종의 체중 분포가 어떻게 다른가?
- **비대각선**: 산점도에서 각 종별 데이터 클러스터를 확인할 수 있습니다.
  - 예: `bill_length_mm` vs `bill_depth_mm` 산점도에서 세 종이 명확히 구분되는가?
  - 만약 구분된다면, 이 두 변수만으로 종 분류가 가능할 수 있습니다.

### 인사이트

hue를 사용하면 **숨겨진 패턴**을 발견할 수 있습니다.
예를 들어, 전체 데이터에서는 두 변수가 선형 관계에 보이지만,
종별로 나누어 보면 각 종 내에서는 관계가 약하거나 반대 방향일 수 있습니다.
이는 **심슨의 역설(Simpson's Paradox)**을 발견하는 데 도움이 됩니다.


````python
# 다변량(Multivariate) — pairplot

# hue + palette
sns.pairplot(data=df_penguins, hue="species", palette="Set2")
plt.show()
````

### pairplot — kind와 diag_kind 상세

`kind`와 `diag_kind`를 조합하여 다양한 시각화를 생성할 수 있습니다.

### kind 옵션 (비대각선)

| 값 | 설명 | 활용 |
|---|---|---|
| `"scatter"` | 산점도 (기본) | 변수 간 관계의 전체적인 패턴 파악 |
| `"kde"` | 2D KDE 등고선 | 데이터 밀도의 집중 영역 파악 |
| `"hist"` | 2D 히스토그램 | 대용량 데이터의 밀도 분포 |
| `"reg"` | 회귀선 + 산점도 | 선형 관계의 강도와 방향 파악 |

### diag_kind 옵션 (대각선)

| 값 | 설명 | 활용 |
|---|---|---|
| `"auto"` | hue 있으면 KDE, 없으면 hist (기본) | 자동 선택 |
| `"hist"` | 히스토그램 | 데이터의 구간별 빈도 |
| `"kde"` | KDE 곡선 | 데이터의 연속적 밀도 분포 |
| `None` | 표시 안 함 | 간결한 표현 |

### 추가 옵션 전달

- **`diag_kws`**: 대각선 그래프에 추가 옵션을 전달합니다.
  - `diag_kind="hist"`일 때: `multiple`, `bins`, `binwidth`, `alpha`, `stat` 등
  - `diag_kind="kde"`일 때: `fill`, `alpha`, `linewidth`, `bw_adjust` 등
- **`plot_kws`**: 비대각선 그래프에 추가 옵션을 전달합니다.
  - `kind="scatter"`일 때: `alpha`, `s`, `edgecolor`, `linewidth` 등
  - `kind="kde"`일 때: `fill`, `alpha`, `linewidth`, `bw_adjust` 등

### 2D KDE 해석

- 등고선이 촘촘할수록 → 해당 영역에 데이터가 많이 몰려 있음
- 등고선이 넓게 퍼질수록 → 데이터가 넓게 분포함
- 중심점(peak): 데이터의 최빈 영역


````python
# 다변량(Multivariate) — pairplot

# kind + diag_kind
# kind: "scatter" (기본) / "kde" / "hist" / "reg"
# diag_kind: "auto" (기본) / "hist" / "kde" / None
sns.pairplot(data=df_penguins, kind="kde", diag_kind="hist", hue="species", diag_kws={"multiple": "dodge"})
plt.show()

# diag_kws는 대각선 그래프에, plot_kws는 비대각선 그래프에 추가 옵션을 전달

# 1. diag_kws의 키는

# diag_kind="hist"일 때 → histplot 파라미터: multiple, bins, binwidth, alpha, stat
# "multiple": "layer", "stack", "dodge", "fill"
# bins: 5, 10, 20, 50 (정수)
# binwidth: 0.5, 1.0, 2.0 (실수)
# alpha: 0.0 ~ 1.0 (실수)
# stat: "count", "frequency", "density", "probability"

# diag_kind="kde"일 때 → kdeplot 파라미터: fill, alpha, linewidth, bw_adjust
# fill: True, False
# alpha: 0.0 ~ 1.0 (실수)
# linewidth: 0.5, 1.0, 2.0 (실수)
# bw_adjust: 0.5 (뾰족), 1.0 (기본), 2.0 (부드러움)


# 2. plot_kws의 키는

# kind="scatter"일 때 → scatterplot 파라미터: alpha, s, edgecolor, linewidth
# alpha: 0.0 ~ 1.0 (실수)
# s: 10, 50, 100 (점 크기, 정수)
# edgecolor: "black", "white", "none" (문자열)
# linewidth: 0.5, 1.0, 2.0 (실수)

# kind="kde"일 때 → kdeplot 파라미터: fill, alpha, linewidth, bw_adjust
# fill: True, False
# alpha: 0.0 ~ 1.0 (실수)
# linewidth: 0.5, 1.0, 2.0 (실수)
# bw_adjust: 0.5 (뾰족), 1.0 (기본), 2.0 (부드러움)

# kind="hist"일 때 → histplot 파라미터: bins, alpha, stat
# bins: 5, 10, 20, 50 (정수)
# alpha: 0.0 ~ 1.0 (실수)
# stat: "count", "frequency", "density", "probability"


# kind="kde"의 경우 비대각 성분이 2D KDE
# 등고선이 촘촘할수록 → 그 영역에 데이터가 많이 몰려 있음
# 등고선이 넓을수록  → 데이터가 넓게 퍼져 있음
````

### PairGrid — 특정 컬럼만 선택

`pairplot`은 기본적으로 모든 수치형 컬럼을 사용하지만,
`vars` 파라미터를 사용하여 **특정 컬럼만 선택**할 수 있습니다.

- **`vars=["bill_length_mm", "flipper_length_mm", "body_mass_g"]`**:
  - 3개의 컬럼만 선택하여 3×3 격자를 생성합니다.
  - 전체 컬럼이 많을 때 관심 있는 변수들만 집중적으로 분석할 수 있습니다.
- **`hue="species"`**: 종별로 색상 구분

### 활용

- 변수가 10개 이상인 대규모 데이터셋에서는 전체 `pairplot`이 너무 복잡해집니다(10×10=100개 그래프).
- `vars`를 사용하여 핵심 변수 3~5개만 선택하면 가독성이 크게 향상됩니다.
- 도메인 지식을 활용하여 분석에 필수적인 변수만 선별하는 것이 좋습니다.

> **참고**: `pairplot` 내부적으로 `PairGrid` 객체를 생성하고 관리합니다.
> `PairGrid`를 직접 사용하면 더 세밀한 커스터마이징이 가능하지만, 사용법이 더 복잡합니다.


````python
# 다변량(Multivariate) — PairGrid

# vars — 특정 컬럼만 선택
sns.pairplot(data=df_penguins,
             vars=["bill_length_mm", "flipper_length_mm", "body_mass_g"],
             hue="species")
plt.show()
````
