## Seaborn 시각화 카테고리 개요

이 노트북에서는 Seaborn의 주요 시각화 유형을 다룹니다. Seaborn은 Matplotlib을 기반으로 한 Python 데이터 시각화 라이브러리로, 통계적 그래프를 쉽게 그릴 수 있게 해줍니다.

### 주요 카테고리

| 카테고리 | 설명 | 주요 함수 |
|---|---|---|
| **분포(Distribution)** | 데이터의 분포 형태를 파악 | `histplot`, `kdeplot`, `ecdfplot`, `rugplot`, `displot` |
| **관계형(Relational)** | 두 변수 간의 관계를 시각화 | `scatterplot`, `lineplot`, `relplot` |
| **범주형(Categorical)** | 범주별 데이터의 분포나 값을 비교 | `barplot`, `countplot`, `boxplot`, `violinplot`, `stripplot`, `swarmplot`, `pointplot`, `catplot` |
| **회귀(Regression)** | 회귀 직선과 잔차 분석 | `regplot`, `lmplot`, `residplot` |
| **행렬(Matrix)** | 데이터 행렬의 패턴을 색상으로 표현 | `heatmap`, `clustermap` |
| **다변량(Multivariate)** | 여러 변수 간의 관계를 한 번에 시각화 | `pairplot`, `PairGrid`, `FacetGrid`, `jointplot` |

각 카테고리는 데이터의 특성과 분석 목적에 따라 적절히 선택하여 사용합니다.


```python
# 분포(Distribution) — histplot, kdeplot, ecdfplot, rugplot, displot
# 관계형(Relational) — scatterplot, lineplot, relplot
# 범주형(Categorical) — barplot, countplot, boxplot, violinplot, stripplot, swarmplot, pointplot, catplot
# 회귀(Regression) — regplot, lmplot, residplot
# 행렬(Matrix) — heatmap, clustermap
# 다변량(Multivariate) — pairplot, PairGrid, FacetGrid, jointplot
```

### 라이브러리 임포트

데이터 시각화에 필요한 핵심 라이브러리를 임포트합니다.

- **`matplotlib.pyplot`**: Python의 기본 시각화 라이브러리인 Matplotlib의 pyplot 모듈입니다. `plt`라는 별칭으로 주로 사용되며, 그래프의 기본 틀(figure, axes)을 생성하고 제어하는 역할을 합니다.
- **`seaborn`**: 통계적 데이터 시각화를 위한 고수준 인터페이스를 제공하는 라이브러리입니다. `sns`라는 별칭으로 사용되며, Matplotlib 기반으로 더 아름답고 통계적으로 의미 있는 그래프를 쉽게 그릴 수 있게 해줍니다.
- **`pandas`**: 데이터 조작 및 분석을 위한 핵심 라이브러리입니다. DataFrame이라는 2차원 테이블 형태의 데이터 구조를 제공하며, Seaborn의 `data` 파라미터에 주로 전달됩니다.

이 세 가지 라이브러리는 Python 데이터 과학 생태계에서 시각화 작업의 표준 조합입니다.

## 다변량(Multivariate) — PairGrid vs pairplot

### PairGrid와 pairplot의 차이

- **`pairplot`**: 고수준 함수로, 한 번의 호출로 대각선과 비대각선에 동일한 그래프를 그립니다. 사용이 간편하지만 세부적인 커스터마이징이 어렵습니다.
- **`PairGrid`**: 저수준 클래스로, **대각선(diag)**, **상삼각(upper)**, **하삼각(lower)** 각 영역에 서로 다른 그래프 함수를 매핑할 수 있습니다. 더 복잡하고 다양한 시각화를 만들 수 있습니다.

### PairGrid의 핵심 메서드

| 메서드 | 설명 |
|---|---|
| `map(func)` | 모든 셀에 동일한 함수 적용 |
| `map_diag(func)` | 대각선 셀에만 함수 적용 |
| `map_offdiag(func)` | 대각선을 제외한 모든 셀에 함수 적용 |
| `map_upper(func)` | 상삼각 영역에 함수 적용 |
| `map_lower(func)` | 하삼각 영역에 함수 적용 |
| `add_legend()` | 범례 추가 |

PairGrid를 사용하면 대칭적인 산점도 행렬에서 중복 정보를 줄이고, 각 영역에 다른 시각화 기법을 적용하여 더 풍부한 정보를 전달할 수 있습니다.


```python
# pairgrid를 간단하게 쓸 수 있게 한 것이 pairplot

# pairgrid는 대각선/상삼각/하삼각을 각각 다르게 지정 가능
```

### PairGrid — 기본 사용법

`PairGrid`의 가장 기본적인 사용 방법입니다.

### 핵심 파라미터

- **`data`**: DataFrame — 여기서는 `df_penguins`

### 작동 원리

1. `sns.PairGrid(df_penguins)`를 호출하면, 펭귄 데이터셋의 수치형 컬럼들 간의 모든 조합에 대한 빈 격자(서브플롯 행렬)가 생성됩니다.
2. `g.map(sns.scatterplot)`를 호출하면, 생성된 모든 셀에 산점도를 그립니다.
3. 대각선 셀에서도 산점도가 그려지므로, `pairplot`의 기본 동작과 유사합니다.

### 결과 해석

- 모든 수치형 변수 간의 산점도를 한 번에 확인할 수 있습니다.
- 대각선은 동일한 변수 간의 산점도이므로 직선이 됩니다(의미 없음).
- 이후 예제에서는 대각선에 히스토그램을, 비대각선에 산점도를 적용하여 더 유용한 시각화를 만들 것입니다.


```python
# 다변량(Multivariate) — PairGrid

# 기본 PairGrid
g = sns.PairGrid(df_penguins)
g.map(sns.scatterplot)
plt.show()
```

### PairGrid — 대각선과 비대각선 분리 (map_diag + map_offdiag)

PairGrid의 가장 일반적인 사용 패턴입니다. 대각선과 비대각선에 서로 다른 그래프를 적용합니다.

### 코드 설명

- **`g.map_diag(sns.histplot)`**: 대각선 셀에 히스토그램을 그립니다.
  - 각 수치형 변수의 **단변량 분포**를 보여줍니다.
  - 예: `bill_length_mm`의 분포는 정규분포에 가까운가?
- **`g.map_offdiag(sns.scatterplot)`**: 대각선을 제외한 모든 셀에 산점도를 그립니다.
  - 두 변수 간의 **이변량 관계**를 보여줍니다.
  - 예: `flipper_length_mm`과 `body_mass_g`는 양의 상관관계가 있는가?

### 장점

- `pairplot`의 기본 동작(diag_kind="hist", kind="scatter")과 동일하지만,
  추가적인 커스터마이징(예: 히스토그램 bins 조정)이 `map_diag` 호출 시 kwargs로 전달되어 가능합니다.


```python
# 다변량(Multivariate) — PairGrid

# map_diag + map_offdiag
g = sns.PairGrid(df_penguins)
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
plt.show()
```

### PairGrid — 상삼각/하삼각 분리 (map_upper + map_lower + map_diag)

PairGrid의 가장 강력한 기능 중 하나는 **대각선, 상삼각, 하삼각을 서로 다르게 지정**할 수 있다는 것입니다.

### 코드 설명

- **`g.map_diag(sns.histplot)`**: 대각선에 히스토그램 (단변량 분포)
- **`g.map_upper(sns.scatterplot)`**: 상삼각에 산점도 (개별 데이터 포인트)
- **`g.map_lower(sns.kdeplot)`**: 하삼각에 2D KDE (밀도 등고선)

### 시각적 효과

- **상삼각(산점도)**: 개별 데이터 포인트를 직접 볼 수 있어 이상치나 특이점을 파악하기 쉽습니다.
- **하삼각(2D KDE)**: 데이터의 밀도 분포를 부드러운 등고선으로 보여주어, 어떤 영역에 데이터가 많이 몰려 있는지 직관적으로 파악할 수 있습니다.
- **대각선(히스토그램)**: 각 변수의 단변량 분포를 보여줍니다.

### 활용

이 조합은 데이터 탐색(EDA) 단계에서 **개별 포인트와 전체 밀도를 동시에** 확인하고 싶을 때 매우 유용합니다.


```python
# 다변량(Multivariate) — PairGrid

# map_upper + map_lower + map_diag
g = sns.PairGrid(df_penguins)
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
plt.show()
```

### PairGrid — hue와 palette

`PairGrid`에도 `hue`와 `palette`를 적용하여 범주형 변수에 따라 데이터를 색상 구분할 수 있습니다.

### 핵심 파라미터

- **`hue="species"`**: 펭귄 종(Adelie, Chinstrap, Gentoo)으로 색상 구분
- **`palette="Set2"`**: 부드러운 색상 조합

### 코드 설명

- **`g.map_diag(sns.histplot)`**: 대각선에 히스토그램 — 각 종별로 다른 색상의 막대가 겹쳐 표시됩니다.
- **`g.map_offdiag(sns.scatterplot)`**: 비대각선에 산점도 — 각 종별로 다른 색상의 점이 표시됩니다.
- **`g.add_legend()`**: 범례를 추가합니다. `PairGrid`는 자동으로 범례를 생성하지 않으므로 수동으로 호출해야 합니다.

### 인사이트

- `bill_length_mm` vs `bill_depth_mm` 산점도에서 세 종이 명확히 구분된다면, 이 두 변수만으로 종 분류가 가능할 수 있습니다.
- 대각선 히스토그램에서 종별 분포가 겹치는 정도를 확인하여, 어떤 변수가 종을 구분하는 데 가장 유용한지 파악할 수 있습니다.


```python
# 다변량(Multivariate) — PairGrid

# hue + palette
g = sns.PairGrid(df_penguins, hue="species", palette="Set2")
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
g.add_legend()
plt.show()
```

### PairGrid — 특정 컬럼만 선택 (vars)

`PairGrid`는 기본적으로 모든 수치형 컬럼을 사용하지만, **`vars` 파라미터**를 사용하여 특정 컬럼만 선택할 수 있습니다.

### 핵심 파라미터

- **`vars=["bill_length_mm", "flipper_length_mm", "body_mass_g"]`**: 3개의 컬럼만 선택하여 3×3 격자를 생성합니다.

### 활용

- 변수가 10개 이상인 대규모 데이터셋에서는 전체 `PairGrid`가 너무 복잡해집니다(10×10=100개 그래프).
- `vars`를 사용하여 핵심 변수 3~5개만 선택하면 가독성이 크게 향상됩니다.
- 도메인 지식을 활용하여 분석에 필수적인 변수만 선별하는 것이 좋습니다.

### PairGrid vs pairplot

- `pairplot` 내부적으로 `PairGrid` 객체를 생성하고 관리합니다.
- `PairGrid`를 직접 사용하면 더 세밀한 커스터마이징이 가능하지만, 사용법이 더 복잡합니다.


```python
# 다변량(Multivariate) — PairGrid

# vars — 특정 컬럼만 선택
g = sns.PairGrid(df_penguins,
                 vars=["bill_length_mm", "flipper_length_mm", "body_mass_g"])
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
plt.show()
```

## 다변량(Multivariate) — FacetGrid

### FacetGrid 개념

`FacetGrid`는 데이터를 **범주형 변수에 따라 여러 서브플롯(패널)로 분할**하여 표시하는 다중 패널 그리드입니다.

- **`relplot`**, **`catplot`**, **`displot`**의 `col`과 `row` 파라미터는 내부적으로 `FacetGrid`를 사용합니다.
- `FacetGrid`를 직접 사용하면 더 세밀한 제어가 가능합니다.

  - 두 번째 패널: 여성(Female)의 팁 분포

### 해석

두 패널을 나란히 비교하여 '남성과 여성 중 누가 더 큰 팁을 주는가?' 또는 '팁 분포의 형태가 성별에 따라 다른가?'를 시각적으로 파악할 수 있습니다.


```python
# 다변량(Multivariate) — FacetGrid

# 기본 FacetGrid (col)
g = sns.FacetGrid(df_tips, col="sex")   # df_tips의 sex 컬럼의 고유값을 기준으로 열을 나눈 빈 subplot을 생성
g.map(sns.histplot, "tip")              # 각 subplot에 x="tip"으로 sns.histplot을 그림
plt.show()
```

### FacetGrid — 열 + 행 분할 (col + row)

`col`과 `row`를 동시에 사용하여 **2차원 패널 그리드**를 생성할 수 있습니다.

### 코드 설명

- **`col="sex"`**: 성별(Male/Female)로 열을 분할 → 2열
- **`row="smoker"`**: 흡연 여부(Yes/No)로 행을 분할 → 2행
- **결과**: 총 2×2 = 4개의 서브플롯이 생성됩니다.
- **`g.map(sns.histplot, "tip")`**: 각 서브플롯에 팁의 히스토그램을 그립니다.

### 각 패널의 의미

| | Male (col) | Female (col) |
|---|---|---|
| **Smoker=Yes (row)** | 남성 흡연자의 팁 분포 | 여성 흡연자의 팁 분포 |
| **Smoker=No (row)** | 남성 비흡연자의 팁 분포 | 여성 비흡연자의 팁 분포 |

### 활용

두 개의 범주형 변수(성별, 흡연 여부)를 동시에 고려하여 팁 분포를 비교할 수 있습니다.
이는 **교차 분석**을 시각적으로 수행하는 것과 같습니다.


```python
# 다변량(Multivariate) — FacetGrid

# col + row
g = sns.FacetGrid(df_tips, col="sex", row="smoker")
# df_tips 데이터의
# sex 컬럼의 고유값을 기준으로 열, smoker 컬럼의 고유값을 기준으로 행을 나눈 2행 2열의 빈 subplot을 생성

g.map(sns.histplot, "tip")
# 4개의 각 subplot에 x="tip"으로 sns.histplot을 그림
plt.show()
```

### FacetGrid — hue와 palette

`FacetGrid`에 `hue`를 추가하면 각 패널 내에서 추가 범주로 데이터를 색상 구분할 수 있습니다.

### 코드 설명

- **`col="sex"`**: 성별로 열 분할
- **`hue="smoker"`**: 흡연 여부로 색상 구분
- **`palette="Set2"`**: 색상 팔레트 적용
- **`g.map(sns.scatterplot, "total_bill", "tip")`**: 각 패널에 x="total_bill", y="tip" 산점도를 그립니다.
- **`g.add_legend()`**: hue에 대한 범례를 추가합니다.

### 범례 대상

`add_legend()`는 `hue`, `style`, `size` 등의 미적 매핑(aesthetic mapping)에 대한 범례를 자동으로 생성합니다.
`FacetGrid`는 `PairGrid`와 마찬가지로 범례를 자동으로 생성하지 않으므로 수동 호출이 필요합니다.

### 해석

남성/여성 패널별로 총액(total_bill) 대비 팁(tip)의 관계를 보면서, 흡연 여부에 따라 점의 색상이 다른 것을 확인할 수 있습니다.


```python
# 다변량(Multivariate) — FacetGrid

# hue + palette
g = sns.FacetGrid(df_tips, col="sex", hue="smoker", palette="Set2")
g.map(sns.scatterplot, "total_bill", "tip")     # 4개의 각 subplot에 x="total_bill", y="tip"으로 sns.scatterplot을 그림
g.add_legend()                                  # legend의 대상은 hue, style, size
plt.show()
```

### FacetGrid — 그래프 함수 변경 (boxplot)

`FacetGrid`는 `map` 메서드의 첫 번째 인자로 **어떤 Seaborn 함수든** 전달할 수 있습니다.

### 코드 설명

- **`g.map(sns.boxplot, "day", "tip", order=["Thur", "Fri", "Sat", "Sun"])`**: 각 패널에 x="day", y="tip"인 박스플롯을 그립니다.
- **`order`**: x축 범주의 표시 순서를 지정합니다.

### 결과

남성(Male)과 여성(Female) 패널별로 요일별 팁의 분포를 박스플롯으로 비교할 수 있습니다.
이를 통해 '성별에 따라 어떤 요일에 팁이 더 많은가?' 또는 '팁 분포의 변동성이 성별/요일에 따라 다른가?'를 파악할 수 있습니다.

### 주의사항

`map`에 전달하는 함수의 추가 파라미터(예: `order`)는 `map` 호출 시 함께 전달됩니다.
이는 `FacetGrid`를 사용한 커스터마이징의 유연성을 보여줍니다.


```python
# 다변량(Multivariate) — FacetGrid

# map 함수를 sns.boxplot로 변경
g = sns.FacetGrid(df_tips, col="sex")
g.map(sns.boxplot, "day", "tip", order=["Thur", "Fri", "Sat", "Sun"])
plt.show()
```

### FacetGrid — 레이아웃 조정 (col_wrap + height + aspect)

`FacetGrid`의 레이아웃을 세밀하게 조정할 수 있습니다.

### 핵심 파라미터

- **`col="day"`**: 요일(Thur, Fri, Sat, Sun)별로 열 분할 → 4개 패널
- **`col_wrap=2`**: 한 행에 최대 2개까지만 배치, 넘으면 줄바꿈
  - 결과: 2행 2열 레이아웃
- **`height=3`**: 각 서브플롯의 높이를 3인치로 설정
- **`aspect=1.2`**: 각 서브플롯의 가로/세로 비율
  - **너비 = height × aspect = 3 × 1.2 = 3.6인치**

### 활용

- `col_wrap`은 패널이 많을 때(예: 12개월) 한 행에 너무 많은 패널이 들어가는 것을 방지합니다.
- `height`와 `aspect`는 각 패널의 크기를 조절하여 가독성을 높입니다.
- 이들을 적절히 조합하면 보고서나 발표용 그래프의 레이아웃을 최적화할 수 있습니다.


```python
# 다변량(Multivariate) — FacetGrid

# col_wrap + height + aspect
g = sns.FacetGrid(df_tips, col="day", col_wrap=2, height=3, aspect=1.2)
# day: 고유값 기준으로 열 방향 subplot 생성 (4개)
# col_wrap=2: 한 행에 최대 2개까지만 배치, 넘으면 줄바꿈
# height=3: 각 subplot의 높이 (인치)
# aspect=1.2: 각 subplot의 가로/세로 비율 (너비 = height × aspect)
# 따라서 너비는 height × aspect =3 x 1.2 = 3.6인치

g.map(sns.histplot, "tip")
plt.show()
```

### FacetGrid — 축 공유 (sharex=True, sharey=True)

`sharex`와 `sharey`를 사용하여 여러 서브플롯 간의 축 범위를 공유할지 여부를 제어할 수 있습니다.

### sharex=True, sharey=True (기본값)

- **모든 서브플롯이 동일한 x축과 y축 범위**를 사용합니다.
- **장점**: 서브플롯 간 직접 비교가 쉽습니다. (같은 눈금 기준으로 비교 가능)
- **단점**: 한 패널에 극단적인 값이 있으면 다른 패널의 세부 변화가 압축되어 보일 수 있습니다.

### 예시 해석

남성(Male)과 여성(Female)의 팁 히스토그램을 동일한 x축/y축 범위로 비교하면,
두 그룹 간 팁 금액의 분포 차이를 더 정확하게 파악할 수 있습니다.


```python
# 다변량(Multivariate) — FacetGrid

# sharex + sharey — 축 공유 비교
# True (기본) → 모든 subplot이 같은 축 범위 사용
g = sns.FacetGrid(df_tips, col="sex", sharex=True, sharey=True)
g.map(sns.histplot, "tip")
plt.show()
```

### FacetGrid — 축 독립 (sharex=False, sharey=False)


`sharex=False`와 `sharey=False`를 설정하면 **각 서브플롯이 독립적인 축 범위**를 사용합니다.

### sharex=False, sharey=False

- **각 서브플롯이 자신의 데이터에 최적화된 축 범위**를 사용합니다.
- **장점**: 각 패널의 세부 분포를 최대한 확대하여 볼 수 있습니다.
- **단점**: 서브플롯 간 직접 비교가 어렵습니다. (눈금 기준이 다름)

### 선택 가이드

| 상황 | 추천 설정 |
|---|---|
| 패널 간 직접 비교가 중요 | `sharex=True`, `sharey=True` |
| 각 패널의 세부 분포가 중요 | `sharex=False`, `sharey=False` |
| x축만 통일하고 y축은 독립 | `sharex=True`, `sharey=False` |
| y축만 통일하고 x축은 독립 | `sharex=False`, `sharey=True` |

데이터의 특성과 분석 목적에 따라 적절한 설정을 선택하는 것이 중요합니다.


```python

# False → 각 subplot이 독립적인 축 범위 사용
g = sns.FacetGrid(df_tips, col="sex", sharex=False, sharey=False)
g.map(sns.histplot, "tip")
plt.show()
```
