## Python datetime vs Pandas Timestamp

### 개념 비교

Python의 표준 라이브러리 `datetime`과 Pandas의 `Timestamp`는 모두 **특정 시점**을 나타내는 데 사용됩니다.

| 특성 | Python `datetime` | Pandas `Timestamp` |
|---|---|---|
| **상속 관계** | 기본 클래스 | `datetime`을 상속받음 |
| **패키지** | `datetime` (표준 라이브러리) | `pandas` |
| **주요 용도** | 날짜/시간 연산 기본 | 시계열 데이터 분석 |
| **인덱스 사용** | 직접 사용 불편 | `DatetimeIndex`로 편리하게 사용 |
| **시간대 지원** | 기본 `timezone` 모듈 | 내장 `tz` 파라미터 |

### 코드 설명

- **`datetime(2024, 6, 4)`**: Python의 datetime 객체 생성 (2024년 6월 4일)
- **`pd.Timestamp('2024-06-04')`**: Pandas의 Timestamp 객체 생성
- **`isinstance(ts, datetime)`**: `Timestamp`가 `datetime`의 서브클래스인지 확인 → `True` 반환

### 왜 Pandas Timestamp를 사용하는가?

- **백터화 연산**: 여러 Timestamp에 대한 연산을 한 번에 수행할 수 있습니다.
- **시계열 기능**: 리샘플링(resample), 이동평균(rolling), 시프트(shift) 등 고급 시계열 기능 제공
- **NaT (Not a Time)**: 누락된 날짜/시간을 효율적으로 처리
- **NaN과 유사한 개념**: NaT는 날짜/시간 데이터의 결측치를 나타냅니다.


````python
# Python에 datetime이 있다면 Pandas에는 Timestamp가 있음
# 둘 다, 특정 시점을 나타내는 데 사용

from datetime import datetime
import pandas as pd

dt = datetime(2024, 6, 4)
ts = pd.Timestamp('2024-06-04')

# Timestamp는 datetime을 상속
isinstance(ts, datetime)  # True
````

### datetime vs Timestamp 메서드 비교

Python `datetime`과 Pandas `Timestamp`는 유사한 기능을 제공하지만, 메서드 이름이 약간 다릅니다.

| 작업 | Python `datetime` | Pandas `Timestamp` |
|---|---|---|
| **문자열 → 객체** | `datetime.strptime()` | `pd.to_datetime()` |
| **객체 → 문자열** | `dt.strftime()` | `ts.strftime()` |

### 메서드 설명

- **`strptime`**: **STRing Parse TIME** — 문자열을 지정된 형식에 따라 날짜/시간 객체로 파싱합니다.
- **`strftime`**: **STRing Format TIME** — 날짜/시간 객체를 지정된 형식의 문자열로 변환합니다.
- **`pd.to_datetime()`**: Pandas의 범용 날짜 파싱 함수 — 다양한 형식의 문자열을 자동으로 인식합니다.

> **팁**: `pd.to_datetime()`은 다양한 형식을 자동으로 인식하므로, 일반적으로 `strptime`보다 더 편리하게 사용할 수 있습니다.


````python
#                   datetime                Timestamp
# 문자열 → 객체      datetime.strptime()      pd.to_datetime()
# 객체 → 문자열      dt.strftime()            ts.strftime()
````

## Timestamp → 문자열 변환 (strftime)

`strftime` 메서드를 사용하여 `Timestamp` 객체를 원하는 형식의 문자열로 변환할 수 있습니다.

### 주요 형식 지정자 (Format Specifiers)

| 지정자 | 설명 | 예시 |
|---|---|---|
| `%Y` | 4자리 연도 | `2024` |
| `%y` | 2자리 연도 | `24` |
| `%m` | 2자리 월 | `06` |
| `%B` | 월 전체 이름 | `June` |
| `%b` | 월 약자 | `Jun` |
| `%d` | 2자리 일 | `04` |
| `%A` | 요일 전체 이름 | `Tuesday` |
| `%a` | 요일 약자 | `Tue` |
| `%H` | 24시간제 시 | `09` |
| `%I` | 12시간제 시 | `09` |
| `%p` | AM/PM | `AM` |
| `%M` | 분 | `30` |
| `%S` | 초 | `45` |
| `%f` | 마이크로초 | `123456` |
| `%j` | 연중 몇 번째 날 | `156` |
| `%W` | 연중 몇 번째 주 | `23` |

### 활용

- 보고서나 파일명에 날짜를 특정 형식으로 포함할 때
- 데이터베이스에 저장할 때 특정 문자열 형식으로 변환할 때
- 사용자에게 날짜를 보여줄 때 원하는 형식으로 포맷팅할 때


````python
# Timestamp → 문자열
````

### strftime 예시 코드

`pd.Timestamp('2024-06-04 09:30:45')` 객체를 다양한 형식의 문자열로 변환하는 예시입니다.

### 예시별 설명

- **`ts.strftime('%Y-%m-%d')`** → `'2024-06-04'`
  - 가장 일반적인 날짜 형식 (ISO 8601 기본 형식)
- **`ts.strftime('%Y/%m/%d %H:%M:%S')`** → `'2024/06/04 09:30:45'`
  - 슬래시 구분자와 시간을 포함한 형식
- **`ts.strftime('%Y년 %m월 %d일')`** → `'2024년 06월 04일'`
  - 한글 날짜 형식 (보고서나 문서용)
- **`ts.strftime('%b %Y')`** → `'Jun 2024'`
  - 월 약자와 연도 (영문 보고서용)

### 포맷 문자열 팁

- 형식 지정자 사이에 `-`, `/`, `.`, 공백, 한글 등 원하는 구분자를 자유롭게 사용할 수 있습니다.
- `%B`는 전체 월 이름(January), `%b`는 약자(Jan)를 반환합니다.


````python
import pandas as pd

ts = pd.Timestamp('2024-06-04 09:30:45')

# strftime: Timestamp를 문자열로 변환

ts.strftime('%Y-%m-%d')           # '2024-06-04'
ts.strftime('%Y/%m/%d %H:%M:%S')  # '2024/06/04 09:30:45'
ts.strftime('%Y년 %m월 %d일')      # '2024년 06월 04일'
ts.strftime('%b %Y')              # 'Jun 2024'

# %b        Jan, Feb, Mar, ... Jun
# %B        January, February, March, ... June

# %Y        4자리 연도      2024
# %y        2자리 연도      24
# %m        2자리 월        06
# %B        월 전체 이름    June
# %b        월 약자        Jun
# %d        2자리 일        04
# %A        요일 전체 이름  Tuesday
# %a        요일 약자       Tue
# %H        24시간제 시     09
# %I        12시간제 시     09
# %p        AM/PM         AM
# %M        분             30
# %S        초             45
# %f        마이크로초      123456
# %j        연중 몇 번째 날 156
# %W        연중 몇 번째 주 23
````

## 문자열 → Timestamp 변환 (pd.to_datetime)

`pd.to_datetime()` 함수는 다양한 형식의 문자열을 Pandas `Timestamp` 객체로 변환합니다.

### 주요 특징

- **자동 형식 인식**: 다양한 날짜/시간 문자열 형식을 자동으로 파싱합니다.
- **벡터화 지원**: 단일 문자열뿐 아니라 리스트, Series 전체를 한 번에 변환할 수 있습니다.
- **`format` 파라미터**: 명시적으로 형식을 지정하여 더 빠르고 정확한 파싱이 가능합니다.

### 기본 사용법

```python
ts = pd.to_datetime('2024-06-04')
print(ts)        # 2024-06-04 00:00:00
print(type(ts))  # <class 'pandas._libs.tslibs.timestamps.Timestamp'>
```

### format 파라미터의 필요성

- 날짜 형식이 모호할 때(예: `04/06/2024`가 4월 6일인지 6월 4일인지) 명확히 지정합니다.
- 형식을 미리 알고 있으면 자동 추론보다 **파싱 속도가 더 빠릅니다**.


````python
# 문자열 → Timestamp
````

````python
ts = pd.to_datetime('2024-06-04')
print(ts)
print(type(ts))
````

### pd.to_datetime 다양한 형식 예시

`pd.to_datetime()`은 다양한 날짜/시간 문자열 형식을 자동으로 인식하고 변환합니다.

### 예시별 설명

- **`pd.to_datetime('2024/06/04')`** → 슬래시 구분자 자동 인식
- **`pd.to_datetime('20240604')`** → 구분자 없는 8자리 날짜 자동 인식 (YYYYMMDD)
- **`pd.to_datetime('June 4, 2024')`** → 영문 월 이름 + 일 + 연도 자동 인식
- **`pd.to_datetime('2024-06-04 09:30:00')`** → 날짜 + 시간 자동 인식
- **`pd.to_datetime('04/06/2024', format='%d/%m/%Y')`**
  - `format` 파라미터로 **명시적으로 형식 지정**
  - `%d/%m/%Y`: 일/월/연도 순서 (유럽식 날짜 형식)
  - 이 형식이 없으면 4월 6일로 잘못 파싱될 수 있습니다 (미국식 MM/DD/YYYY로 추론).
- **`pd.to_datetime('2024년 06월 04일', format='%Y년 %m월 %d일')`**
  - 한글 날짜 형식을 명시적으로 지정하여 파싱

### 에러 처리

```python
# errors='coerce': 파싱 실패 시 NaT(Not a Time)으로 처리
pd.to_datetime(['2024-06-04', 'invalid_date'], errors='coerce')
```


````python
ts1 = pd.to_datetime('2024/06/04')
ts2 = pd.to_datetime('20240604')
ts3 = pd.to_datetime('June 4, 2024')
ts4 = pd.to_datetime('2024-06-04 09:30:00')
ts5 = pd.to_datetime('04/06/2024', format='%d/%m/%Y')               # format으로 형식을 명시적으로 지정
ts6 = pd.to_datetime('2024년 06월 04일', format='%Y년 %m월 %d일')     # format으로 형식을 명시적으로 지정

print(ts1)
print(ts2)
print(ts3)
print(ts4)
print(ts5)
print(ts6)
````

## DatetimeIndex

### DatetimeIndex 개념

`DatetimeIndex`는 **Timestamp 객체들의 인덱스**입니다. Pandas의 DataFrame이나 Series에서 행(또는 열)을 날짜/시간으로 식별할 때 사용됩니다.

### 특징

- **`pd.DatetimeIndex([...])`**: Timestamp 문자열 리스트로 생성
- **시계열 데이터의 인덱스로 사용**: 날짜/시간 기반 데이터 조회, 슬라이싱, 리샘플링 등에 필수적
- **속성 접근**: `.year`, `.month`, `.day` 등으로 날짜 구성 요소에 쉽게 접근

### 활용

- 주식 가격, 기상 데이터, 웹 트래픽 등 시간에 따른 변화를 다루는 데이터의 인덱스로 사용
- `pd.date_range()`로 생성된 날짜 범위도 `DatetimeIndex`를 반환합니다.


````python
# pd.DatetimeIndex는 DataFrame/Series의 인덱스로 사용
````

### DatetimeIndex — 기본 속성 접근

`DatetimeIndex`의 각 Timestamp 요소들은 **벡터화된 속성**을 제공합니다.

### 코드 설명

```python
idx = pd.DatetimeIndex(['2024-01-01', '2024-06-04', '2024-12-31'])
```

- **`idx.year`** → `Index([2024, 2024, 2024])`
  - 각 Timestamp의 연도를 반환합니다.
- **`idx.month`** → `Index([1, 6, 12])`
  - 각 Timestamp의 월을 반환합니다. (1~12)
- **`idx.day`** → `Index([1, 4, 31])`
  - 각 Timestamp의 일을 반환합니다. (1~31)

### 반환 타입

각 속성은 `Index` 객체를 반환하므로, 조건 필터링이나 그룹화에 바로 사용할 수 있습니다.

```python
# 2024년 6월 이후의 데이터만 선택
df[idx.month >= 6]
```


````python
idx = pd.DatetimeIndex(['2024-01-01', '2024-06-04', '2024-12-31'])

print(idx)
print(idx.year)   # Index([2024, 2024, 2024])
print(idx.month)  # Index([1, 6, 12])
print(idx.day)    # Index([1, 4, 31])
````

### DatetimeIndex — 시간 관련 속성

시간 정보가 포함된 Timestamp에 대해서도 벡터화된 속성 접근이 가능합니다.

### 코드 설명

```python
idx = pd.DatetimeIndex(['2024-01-01', '2024-06-04 09:30:45', '2024-12-31 23:59:59'])
```

- **`idx.hour`** → `Index([0, 9, 23])`
  - 24시간제 시 (0~23)
- **`idx.minute`** → `Index([0, 30, 59])`
  - 분 (0~59)
- **`idx.second`** → `Index([0, 45, 59])`
  - 초 (0~59)

### 활용

- 시간대별 분석: 예를 들어, 시간대별 웹사이트 방문량, 시간대별 교통량 등
- 조건 필터링: 예를 들어, 오전(0~11시) 데이터만 선택

```python
# 오전 데이터만 선택
df[idx.hour < 12]
```


````python
idx = pd.DatetimeIndex(['2024-01-01', '2024-06-04 09:30:45', '2024-12-31 23:59:59'])

print(idx.hour)    # Index([0, 9, 23], dtype='int32')
print(idx.minute)  # Index([0, 30, 59], dtype='int32')
print(idx.second)  # Index([0, 45, 59], dtype='int32')
````

### DatetimeIndex — 요일 속성

Timestamp의 요일 정보를 확인할 수 있습니다.

### 코드 설명

```python
idx = pd.DatetimeIndex(['2024-01-01', '2024-06-04', '2024-12-31'])
```

- **`idx.dayofweek`** → `Index([0, 1, 1])`
  - **0=월요일, 1=화요일, 2=수요일, 3=목요일, 4=금요일, 5=토요일, 6=일요일**
  - Python의 `datetime.weekday()`와 동일한 규칙
- **`idx.day_name()`** → `Index(['Monday', 'Tuesday', 'Tuesday'])`
  - 요일의 전체 영문 이름을 반환합니다.
  - 한글 요일명이 필요하면 `.day_name(locale='ko_KR')`을 사용할 수 있습니다.

### 활용

- 주말/평일 구분: `idx.dayofweek >= 5` (토요일=5, 일요일=6)
- 특정 요일 데이터 필터링: `idx.dayofweek == 0` (월요일만)


````python
idx = pd.DatetimeIndex(['2024-01-01', '2024-06-04', '2024-12-31'])

print(idx.dayofweek)   # Index([0, 1, 1], dtype='int32')  0=월요일
print(idx.day_name())  # Index(['Monday', 'Tuesday', 'Tuesday'])
````

### DatetimeIndex — 분기 속성

Timestamp가 속한 **분기(Quarter)**를 확인할 수 있습니다.

### 코드 설명

```python
idx = pd.DatetimeIndex(['2024-01-01', '2024-06-04', '2024-12-31'])
```

- **`idx.quarter`** → `Index([1, 2, 4])`
  - 1분기: 1월~3월
  - 2분기: 4월~6월
  - 3분기: 7월~9월
  - 4분기: 10월~12월

### 활용

- 분기별 매출/실적 집계
- 분기별 데이터 그룹화 및 비교

```python
# 분기별 평균 매출 계산
df.groupby(idx.quarter)['sales'].mean()
```


````python
idx = pd.DatetimeIndex(['2024-01-01', '2024-06-04', '2024-12-31'])

print(idx.quarter)  # Index([1, 2, 4], dtype='int32')
````

### DatetimeIndex — 연중 일수 속성

Timestamp가 **연중 몇 번째 날**인지 확인할 수 있습니다.

### 코드 설명

```python
idx = pd.DatetimeIndex(['2024-01-01', '2024-06-04', '2024-12-31'])
```

- **`idx.dayofyear`** → `Index([1, 156, 366])`
  - `2024-01-01` → 1번째 날
  - `2024-06-04` → 156번째 날
  - `2024-12-31` → 366번째 날 (2024년은 윤년이므로 366일)

### 활용

- 연중 진행률 계산
- 계절별 분석 (예: 봄=dayofyear 80~171)


````python
idx = pd.DatetimeIndex(['2024-01-01', '2024-06-04', '2024-12-31'])

print(idx.dayofyear)  # Index([1, 156, 366], dtype='int32')
````

## pd.date_range() — 날짜 범위 생성

`pd.date_range()` 함수는 **규칙적인 간격으로 날짜 범위를 생성**하여 `DatetimeIndex`를 반환합니다.

### 핵심 파라미터

| 파라미터 | 설명 | 예시 |
|---|---|---|
| `start` | 시작 날짜 | `'2024-01-01'` |
| `end` | 종료 날짜 | `'2024-12-31'` |
| `periods` | 생성할 개수 | `12` |
| `freq` | 빈도(간격) | `'D'`, `'W'`, `'ME'`, `'YE'` |

### freq 주요 값

| 값 | 설명 |
|---|---|
| `'D'` | 일별 (Day) |
| `'W'` | 주별 (Week, 일요일 기준) |
| `'ME'` | 월말 (Month End) |
| `'MS'` | 월초 (Month Start) |
| `'QE'` | 분기말 (Quarter End) |
| `'YE'` | 연말 (Year End) |
| `'YS'` | 연초 (Year Start) |

### 코드 설명

- **`pd.date_range('2024-01-01', periods=12, freq='ME')`**
  - 2024년 1월부터 시작하여 월말 기준으로 12개의 날짜를 생성
  - 결과: 2024년 1월~12월의 월말 날짜들
- **`idx[0:6]`**: 앞에서 6개만 슬라이싱 (1월~6월)

### 활용

- 시계열 데이터의 빈 인덱스 생성
- 월별/분기별/연도별 보고서 기간 설정
- 데이터 리샘플링의 타겟 인덱스 생성


````python
# pd.date_range()의 리턴은 DatetimeIndex

idx = pd.date_range('2024-01-01', periods=12, freq='ME')
# pd.date_range()는 날짜 범위를 생성
# '2024-01-01'      시작 날짜
# periods=12        12개 생성
# freq='ME'         월말(Month End) 기준
# 'D': 일별, 'W': 주별, 'ME': 월말, 'QE': 분기말, 'YE': 연말

print(idx)
print(idx[0:6])  # 앞에서 6개  # 1월 ~ 6월 슬라이싱
````

```python

                   index=pd.date_range('2024-01-01', periods=6, freq='ME'))
```

- **`index=pd.date_range(...)`**: DataFrame의 행 인덱스를 `DatetimeIndex`로 설정합니다.
- **결과**: 2024년 1월~6월 월말을 인덱스로 하는 6개 행의 DataFrame

### DatetimeIndex 인덱스의 장점

1. **날짜 문자열 슬라이싱**: `df['2024-01':'2024-03']`처럼 문자열로 직접 슬라이싱 가능
2. **리샘플링**: `df.resample('QE').mean()`처럼 분기/연도 등으로 쉽게 집계

````python
df = pd.DataFrame({
    '값': [10, 20, 30, 40, 50, 60]
}, index=pd.date_range('2024-01-01', periods=6, freq='ME'))

print(df)
````

## Timedelta — 시간 간(間) 계산

### Timedelta 개념

`Timedelta`는 **두 시점 간의 차이(시간 간격)**를 나타내는 Pandas 객체입니다.

### Python datetime vs Pandas Timedelta

| 특성 | Python `timedelta` | Pandas `Timedelta` |
|---|---|---|
| **패키지** | `datetime` (표준 라이브러리) | `pandas` |
| **기능** | 기본적인 시간 차이 계산 | 고급 시계열 기능 (NaT 지원, 벡터화 등) |
| **단위** | 일, 초, 마이크로초 | 일, 시, 분, 초, 밀리초, 마이크로초, 나노초 |

### 활용

- 두 날짜 간의 경과 일수 계산
- 날짜에 일정 기간을 더하거나 빼기
- 시계열 데이터의 시간 간격 분석


````python
# pd.Timedelta
````

### Timedelta 생성

`pd.Timedelta`는 다양한 방식으로 생성할 수 있습니다.

### 생성 방법

- **`pd.Timedelta('3 days')`** → 문자열로 생성
- **`pd.Timedelta('3 days 2 hours 30 minutes')`** → 복합 단위 문자열
- **`pd.Timedelta(days=3, hours=2, minutes=30)`** → 키워드 인자로 생성

### 지원하는 단위

`weeks`, `days`, `hours`, `minutes`, `seconds`, `milliseconds`, `microseconds`, `nanoseconds`

### 주의사항

- `pd.Timedelta('3 days 2 hours 30 minutes')`에서 단위는 복수형(s)으로 써야 합니다.
- `days=3`처럼 키워드 인자를 사용하면 단수형도 가능합니다.


````python
print(pd.Timedelta('3 days'))
print(pd.Timedelta('3 days 2 hours 30 minutes'))
print(pd.Timedelta(days=3, hours=2, minutes=30))
````

### Timedelta 속성

`Timedelta` 객체의 주요 속성들을 살펴봅니다.

### 코드 설명

- **`td.days`** → `3`
  - **날짜(일) 부분만** 반환합니다.
  - `3 days 2 hours 30 minutes`에서 `3`만 반환합니다.
- **`td.seconds`** → `9000`
  - **날짜를 제외한 나머지 시간 부분**만 초로 반환합니다.
  - `2 hours 30 minutes = 9000 seconds`
  - `days`는 포함되지 않습니다!
- **`td.total_seconds()`** → `270600.0`
  - **전체 시간 간격을 초로 변환**합니다.
  - `3 days (259200s) + 2 hours (7200s) + 30 minutes (1800s) = 270600 seconds`

### 중요한 차이점

| 속성 | 반환 값 | 설명 |
|---|---|---|
| `.days` | `3` | 날짜 부분만 |
| `.seconds` | `9000` | 날짜 제외 시간만 초로 |
| `.total_seconds()` | `270600.0` | 전체를 초로 |

### 활용

- 전체 경과 시간을 초/분/시간 단위로 환산할 때 `.total_seconds()` 사용
- 날짜 부분만 추출할 때 `.days` 사용


````python
td = pd.Timedelta('3 days 2 hours 30 minutes')

print(td.days)         # 3

# seconds는 날짜를 제외한 나머지 시간 부분만 초로 반환
print(td.seconds)      # 9000  (2시간 30분 = 9000초)

# 날자를 포함한 전체 시간을 초로 바꾸려면 total_seconds()
print(td.total_seconds())  # 270600.0  (전체 초)
````

### Timedelta 산술 연산

`Timedelta`는 `Timestamp`와의 덧셈/뺄셈, 그리고 `Timestamp`끼리의 뺄셈에 사용됩니다.

### 코드 설명

- **`ts + td`** → `2024-01-11 00:00:00`
  - Timestamp에 Timedelta를 **더하면** 미래 날짜를 계산합니다.
  - `2024-01-01` + `10 days` = `2024-01-11`
- **`ts - td`** → `2023-12-22 00:00:00`
  - Timestamp에 Timedelta를 **빼면** 과거 날짜를 계산합니다.
  - `2024-01-01` - `10 days` = `2023-12-22`
- **`pd.Timestamp('2024-06-04') - pd.Timestamp('2024-01-01')`** → `155 days`
  - **Timestamp끼리 빼면 Timedelta**가 반환됩니다.
  - 두 날짜 간의 경과 일수를 계산할 수 있습니다.

### 활용

- 마감일 계산: `today + pd.Timedelta(days=30)`
- 회원 가입 후 경과 일수: `today - join_date`
- 프로젝트 기간: `end_date - start_date`


````python
ts = pd.Timestamp('2024-01-01')
td = pd.Timedelta('10 days')

print(ts + td)  # 2024-01-11 00:00:00
print(ts - td)  # 2023-12-22 00:00:00

# Timestamp끼리 빼면 Timedelta
print(pd.Timestamp('2024-06-04') - pd.Timestamp('2024-01-01'))  # 155 days
````

### Timedelta 비교 연산

`Timedelta` 객체들 간의 크기 비교가 가능합니다.

### 코드 설명

- **`td1 > td2`** → `False`
  - `3 days`는 `5 days`보다 크지 않습니다.
- **`td1 < td2`** → `True`
  - `3 days`는 `5 days`보다 작습니다.
- **`td1 == td2`** → `False`
  - 두 Timedelta는 같지 않습니다.

### 활용

- 조건문에서 시간 간격을 비교하여 처리 로직 분기
- 예: `if elapsed > pd.Timedelta(days=7): print('7일 이상 경과')`

### Timedelta의 유용성

- **일정 기간 반복 작업**: `for i in range(10): date = start + pd.Timedelta(days=i*7)`
- **시간 간격 필터링**: `df[df['duration'] > pd.Timedelta(hours=1)]`
- **시계열 데이터 시프트**: `df.index + pd.Timedelta(days=1)`

Timedelta는 시계열 데이터 처리에서 필수적인 도구입니다.


````python
td1 = pd.Timedelta('3 days')
td2 = pd.Timedelta('5 days')

print(td1 > td2)   # False
print(td1 < td2)   # True
print(td1 == td2)  # False
````
